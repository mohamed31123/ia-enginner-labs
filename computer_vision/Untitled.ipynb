```python
import numpy as np 
import matplotlib.pyplot as plt


img = np.array([
    [25, 100, 220, 220, 200, 150],
    [200, 30,  30,  30, 220, 100],
    [220, 200,  30,  30, 100, 150],
    [150, 150, 150, 200, 150, 150],
    [100, 200,  30, 200, 150, 150],
    [100, 100, 100,  25,  25,  25]
], dtype=np.float64)
print(img)

plt.imshow(img , cmap="gray")
plt.axis('off')

I_min = img.min()

I_max = img.max()

enh = (img - I_min)*255/(I_max - I_min)
print(enh)


print("the image after the enharecement")
plt.imshow(enh , cmap="gray")

# Apply a 3x3 mean filter
padded = np.pad(enh, pad_width=1, mode="edge")
filtered = np.zeros_like(enh)

for i in range(enh.shape[0]):
    for j in range(enh.shape[1]):
        window = padded[i:i+3, j:j+3]
        filtered[i, j] = np.mean(window)

print("Image after 3x3 mean filtering:")
print(np.round(filtered, 2))

# Display the images
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(enh, cmap="gray", vmin=0, vmax=255)
plt.title("Enhanced Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(filtered, cmap="gray", vmin=0, vmax=255)
plt.title("3x3 Mean Filter")
plt.axis("off")

plt.tight_layout()
plt.show()


```
