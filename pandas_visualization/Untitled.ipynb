```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


import pandas as pd

data = {
    "Delivery_ID": [1, 2, 3, 4, 5],
    "Distance_km": [2, 5, 3, 10, 4],
    "Time_min": [15, 25, 18, 50, 22],
    "Price_DH": [12, 20, 15, 35, 18]
}

df = pd.DataFrame(data)

print(df.head())
print(df.info())
print(df.describe)
print( 
     "the mean of the Time is : " ,df[ "Time_min"].mean() ,
     "The std is : " , 
      df[ "Time_min"].std(), 
      "The variation is : " , df[ "Time_min"].var() 
     )
print("the shape of df is : " , df.shape)
```
