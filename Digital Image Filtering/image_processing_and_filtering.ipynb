2. Explain the operating principle of the Gaussian filter. What is its purpose?
2. Gaussian Filter: Operating Principle and Purpose
1. What is a Gaussian filter?
A Gaussian filter is a spatial filtering technique used in digital image processing to smooth an image and reduce noise.

It works by replacing each pixel's value with a weighted average of its neighboring pixels.

The weights are calculated using a Gaussian (bell-shaped) mathematical function.
2. How does it work?
The operating principle consists of four steps:
1. Select a kernel: Choose a small matrix, for example, \(3\times3\).
2. Calculate Gaussian weights: Pixels near the center receive higher weights, while distant pixels receive lower weights.
3. Apply convolution: Move the kernel across the image.
4. Calculate new pixel values: Multiply each neighboring pixel by its corresponding weight and sum the results.

Draw the histogram of image (I)

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Matrice de l'image I (8x8)
I = np.array([
    [13, 13, 12, 12, 12, 11, 11, 11],
    [13, 12, 12, 12, 11, 11, 11, 10],
    [12, 12,  8,  7,  6,  5, 10, 10],
    [12, 12,  7,  6,  5,  4, 10, 10],
    [12, 11,  6,  5,  4,  3, 10,  9],
    [11, 11,  5,  4,  3,  2,  9,  9],
    [11, 11, 10, 10, 10,  9,  9,  9],
    [11, 10, 10, 10,  9,  9,  9,  9]
])

# 2. Calcul des occurrences pour chaque niveau de gris
values, counts = np.unique(I, return_counts=True)

# 3. Tracé de l'histogramme
plt.figure(figsize=(8, 4))
plt.bar(values, counts, color='skyblue', edgecolor='black', width=0.7)
plt.xlabel('Niveau de gris (k)', fontsize=12)
plt.ylabel('Nombre de pixels (n_k)', fontsize=12)
plt.title('Histogramme de l\'image (I)', fontsize=14, fontweight='bold')
plt.xticks(np.arange(0, 16))
plt.yticks(np.arange(0, max(counts) + 2, 2))
plt.grid(axis='y', linestyle='--', alpha=0.7)



plt.tight_layout()
plt.show()
```
