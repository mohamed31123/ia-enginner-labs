```python
import numpy as np
import matplotlib.pyplot as plt

image = np.array([
    [6, 5, 4, 3, 2, 2, 3, 4, 5, 6],
    [5, 4, 3, 2, 1, 1, 2, 3, 4, 5],
    [4, 3, 2, 1, 14, 14, 1, 2, 3, 4],
    [3, 2, 1, 14, 12, 12, 0, 1, 2, 3],
    [2, 1, 14, 12, 0, 8, 12, 14, 0, 2],
    [2, 15, 14, 12, 8, 15, 12, 14, 1, 2],
    [3, 2, 1, 15, 12, 12, 14, 1, 2, 3],
    [4, 3, 2, 1, 14, 14, 1, 2, 3, 4],
    [5, 4, 3, 2, 1, 1, 2, 3, 15, 5],
    [6, 5, 4, 3, 2, 2, 3, 4, 5, 6]
])

print("the image shape is " ,image.shape)
plt.imshow(image, cmap="gray", vmin=0, vmax=15)
plt.colorbar(label="Pixel intensity")
plt.title("Original Image")
plt.show()


```

Apply the mean filter on given positions : 

```python
import numpy as np


positions = [
    (2, 2), (4, 7), (5, 5), (5, 9),
    (6, 2), (6, 6), (7, 4), (9, 9)
]

# Keep the filtered results separately
filtered_image = image.copy()

for row, col in positions:

    
    r = row - 1
    c = col - 1

    
    neighborhood = image[r-1:r+2, c-1:c+2]

    # Calculate the mean
    mean_value = np.mean(neighborhood)

    # Replace the central pixel
    filtered_image[r, c] = mean_value

    
plt.imshow(filtered_image ,cmap="gray" ,vmin=0 , vmax=15)
plt.colorbar(label="Pixel intensity")
plt.title("Filtered Image")
plt.show()
```

Apply the median filter on given positions

```python
median_filtered_image = image.copy()

for row, col in positions:

    
    r = row - 1
    c = col - 1

  
    neighborhood = image[r-1:r+2, c-1:c+2]

    median_value = np.median(neighborhood)
    median_filtered_image[r, c] = median_value
plt.imshow(median_filtered_image , cmap="gray",vmin=0 , vmax=14)
plt.colorbar(label="Pixel Intensity")
plt.title("Filtered Image")
plt.show()
```
