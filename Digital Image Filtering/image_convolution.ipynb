```python
import numpy as np

I = np.array([
    [0, 10, 20],
    [30, 20, 10],
    [40, 30, 40]
])

F = np.array([
    [-1,  0, 1],
    [-2,  0, 2],
    [-1,  0, 1]
])


P = np.pad(I, pad_width=1, mode='constant', constant_values=0)

print(P)
```

```python
# Rotate filter by 180 degrees
F_rotated = np.flip(F)
print(F_rotated)
# Create output matrix
output = np.zeros((3,3))
print(output)

# Slide the filter over the padded image
for i in range(I.shape[0]):
    for j in range(I.shape[1]):

        # Extract 3x3 neighborhood
        region = P[i:i+3, j:j+3]

        # Multiply corresponding values and sum
        output[i, j] = np.sum(region * F_rotated)

print("Convolution result:")
print(output)
```
