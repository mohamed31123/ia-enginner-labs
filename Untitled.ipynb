```python
for score in [12, 15, 18]:
    print(score)
```

```python
def calculate_mean(values):
    return sum(values) / len(values)

scores = [12, 15, 18]
print(calculate_mean(scores))  # 15.0
```

List comperhension

```python
scores = [12, 8, 17, 9, 15]

passed_scores = [score for score in scores if score >= 10]

print(passed_scores)  # [12, 17, 15]
```

Understand indexing

```python
scores = [13,14,15,15,14,18]

#Access the first element 
print(scores[0])
#Access the last element
print(scores[-1])

print(scores[::2])
print(scores[::-1])
```
